# Warm-up

```python
from ambdes import ambsys, SimConfig, run_warm_up_audit, plot_warm_up
```

```python
ambsys_data = ambsys(
    csv_path="data/AmbSYS-to-Mar-2026-UI7FG.csv",
    org_code="RYF",
    month="3",
    year=2026,
)

config = SimConfig(
    ambsys_data=ambsys_data,
    warm_up_period=1000,
    data_collection_period = 2000,
)
audit = run_warm_up_audit(config=config, interval=30, n_reps=5)
audit.head(20)
```

```python
plot_warm_up(audit=audit, metric="response_time", category="C1")
```
