# Warm-up

```python
from ambdes import ArrivalConfig, SimConfig, run_warm_up_audit, plot_warm_up
```

```python
arrival_config = ArrivalConfig(arrival_df="data/arrivals.csv")
config = SimConfig(
    arrival_config=arrival_config,
    warm_up_period=1000,
    data_collection_period = 2000,
)
audit = run_warm_up_audit(config=config, interval=30, n_reps=5)
audit.head(20)
```

```python
plot_warm_up(audit=audit, metric="response_time", category="C1")
```
