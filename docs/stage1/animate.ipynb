# Animation for stage 1 model

```python
from ambdes import ArrivalConfig, Model, SimConfig, generate_animation
```

```python
arrival_config = ArrivalConfig(arrival_df="data/arrivals.csv")
config = SimConfig(
    arrival_config=arrival_config,
    warm_up_period=0,
    data_collection_period=60 * 12,  # 12 hours
)
model = Model(run_number=0, config=config)
model.run()
```

```python
generate_animation(model)
```
