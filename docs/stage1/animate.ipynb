# Animation for stage 1 model

```python
from ambdes import Model, SimConfig, ambsys, generate_animation
```

```python
ambsys_data = ambsys(
    csv_path="data/AmbSYS-to-Mar-2026-UI7FG.csv",
    org_code="RYF",
    month="3",
    year=2026,
)
ambsys_data
```

```python
config = SimConfig(
    ambsys_data=ambsys_data,
    run_length=60 * 12,  # 12 hours
)
model = Model(run_number=0, config=config)
model.run()
```

```python
generate_animation(model)
```
