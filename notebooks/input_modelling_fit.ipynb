# Input modelling

## Set-up

```python
from ambdes import FitDist
import pandas as pd
```

```python
TIME_DATA_PATH = "../data/times.csv"
SEED = 42
```

```python
time_data = pd.read_csv(TIME_DATA_PATH)

# The dataset is very large, so we take a sample for this analysis
if len(time_data) > 100_000:
    time_data_sample = time_data.sample(n=100_000, random_state=SEED)
else:
    time_data_sample = time_data.copy()
```

## Mobilisation

```python
fitter = FitDist(metric="C0001_Mobilisation_Duration", df=time_data_sample)
fitter.fit_and_compare(xmax=200)
```

## Time to scene

```python
fitter = FitDist(
    metric="C0004_MobilisationToScene_Duration", df=time_data_sample
)
fitter.fit_and_compare(xmax=1000)
```

## On scene time

```python
fitter = FitDist(metric="C0007_OnScene_Duration", df=time_data_sample)
fitter.fit_and_compare(xmax=8000)
```

## Time to destination (hospital or non-ED)

```python
fitter = FitDist(metric="C0172_SceneToDestination_Duration", df=time_data_sample)
fitter.fit_and_compare(xmax=6000)
```

## Handover time

```python
fitter = FitDist(metric="C0012_Handover_Duration", df=time_data_sample)
fitter.fit_and_compare(xmax=10_000)
```

## Wrap-up time

```python
fitter = FitDist(metric="C0171_WrapUp_Duration", df=time_data_sample)
fitter.fit_and_compare(xmax=2000)
```
